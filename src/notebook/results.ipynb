```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
label = ['KO', 'PEP', 'NVDA','KSU']
id = 'c8zawdg6'
path_storage = f'storage/{id}'
```

```python
for epoch in range(7):

    with open(f'{path_storage}/epoch={epoch}/samples.npy', 'rb') as f:
        samples = np.load(f)

    nrows, ncols = 3, 3
    indices = np.random.choice(len(samples), nrows*ncols, replace=False)

    fig, axis = plt.subplots(nrows, ncols, figsize=(8, 5))
    axis = axis.ravel()

    add_label = True
    for i, ax in zip(indices, axis):
        ax.set_title(i)
        ax.plot(samples[i] / samples[i, 0], label=label if add_label else None)
        add_label = False

    fig.suptitle(f'Epoch: {epoch}')
    fig.legend(ncols=4, loc='upper center', bbox_to_anchor = [0.5, 1.07])
    fig.tight_layout()
    plt.show()
    plt.close(fig)
```
