```python
from typing import Optional
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
def sine_data_generation(n_samples: int, seq_len: int, n_features: int, seed: Optional[int] = None) -> np.ndarray:
    np.random.seed(seed)
    data = []
    for _ in range(n_samples):
        temp = []
        for k in range(n_features):
            freq = np.random.uniform(0, 0.1)
            phase = np.random.uniform(0, 0.1)
            temp_data = [np.sin(freq * j + phase) for j in range(seq_len)]
            temp.append(temp_data)
        temp = np.transpose(np.asarray(temp))
        temp = (temp + 1) * 0.5
        data.append(temp)

    return np.array(data)
```

```python
seq_len, n_features = 24, 5
data_train = sine_data_generation(100000, seq_len, n_features, 0)
data_val = sine_data_generation(5000, seq_len, n_features)
data_train.shape, data_val.shape
```

```python
i = np.random.randint(0, len(data_train))
plt.plot(data_train[i])
plt.show()
plt.close()
```

```python
os.makedirs('data/sine', exist_ok=True)
np.save('data/sine/train.npy', data_train)
np.save('data/sine/val.npy', data_val)
```
