```python
from typing import Optional
import numpy as np
import matplotlib.pyplot as plt
import os
from tqdm import tqdm
```

```python
def ar_gaussian_generation(
    n_samples: int, seq_len: int, n_features: int, phi: float, sigma: float, seed: Optional[int] = None
) -> np.ndarray:
    np.random.seed(seed)

    mean_val = np.zeros([n_features])
    cov_val = np.ones([n_features, n_features]) * sigma
    np.fill_diagonal(cov_val, 1)

    # Generate autoregressive gaussian data
    data = []
    for _ in tqdm(range(n_samples), desc="Dataset generation"):
        temp = np.zeros([seq_len, n_features])
        for k in range(seq_len):
            # Starting feature
            if k == 0:
                temp[k, :] = np.random.multivariate_normal(mean_val, cov_val, 1)

            # AR(1) Generation
            else:
                temp[k, :] = phi * temp[k - 1, :] + (1 - phi) * 1 * np.random.multivariate_normal(mean_val, cov_val, 1)

        data.append(temp)

    return np.array(data)
```

```python
seq_len, n_features = 24, 5

phi = .8
for sigma in [.2, .5, .8]:
    print(f"Generating data with phi={phi}, sigma={sigma}")
    data_train = ar_gaussian_generation(100000, seq_len, n_features, phi, sigma, 0)
    data_val = ar_gaussian_generation(5000, seq_len, n_features, phi, sigma)
    folder = f'data/ar_gaussian/phi={phi}_sigma={sigma}'
    os.makedirs(folder, exist_ok=True)
    np.save(f'{folder}/train.npy', data_train)
    np.save(f'{folder}/val.npy', data_val)

sigma = .8
for phi in [.2, .5]:
    print(f"Generating data with phi={phi}, sigma={sigma}")
    data_train = ar_gaussian_generation(100000, seq_len, n_features, phi, sigma, 0)
    data_val = ar_gaussian_generation(5000, seq_len, n_features, phi, sigma)
    folder = f'data/ar_gaussian/phi={phi}_sigma={sigma}'
    os.makedirs(folder, exist_ok=True)
    np.save(f'{folder}/train.npy', data_train)
    np.save(f'{folder}/val.npy', data_val)
```
